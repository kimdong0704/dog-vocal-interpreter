# Body.py test on Video input

```python
import numpy as np
from PIL import Image
from woofdecoder.body import BodyBranch
from woofdecoder.preprocess.segments import BarkSegment
from woofdata.path import get_data_dir

root = get_data_dir() / "raw"
paths = sorted((root / "rf_dog_emotion").glob("*/YT_ID_220_mp4-*.jpg"))[:4]
frames = [np.array(Image.open(p).convert("RGB")) for p in paths]
print(len(frames), "frames")

segment = BarkSegment(start_s=0.0, end_s=1.0, frames=frames,
                      frame_times_s=[i * 0.25 for i in range(len(frames))])
body = BodyBranch()
out = body.predict(segment)
print(out.present, out.as_dict())
print(out.extras)
```
