# 10 Preprocessing Check

**Purpose:** Check the bark-detector task: its sources, row counts per split, and one loaded sample.

**Inputs:** The prepared manifest and the PlayPanT val audio.

**Outputs:** Printed checks only; no files are written.

**Splits used:** train, val, test row counts are printed (test: counted, not scored); one val sample is loaded.

```python
from woofdata.loader import load_sample
from woofdata.tasks import get_task, task_rows
from woofdecoder.preprocess.segments import BarkSegment
```

```python
TASK = "bark_detector"
NEW_FIELDS = ["sed_class", "confidence", "dog_boxes"]
```

## Check task and rows count

Print the `BarkSegment` fields and the row count per split and source.

```python
# Elements in BarkSegment object
fields = list(BarkSegment.__dataclass_fields__)
print(f"BarkSegment has {len(fields)} fields: {fields}\n")

# Sources used by the task Bark Detector
sources = get_task(TASK).sources
print(f"Task '{TASK}' uses these sources: {list(sources)}\n")

for split in ["train", "val", "test"]:
    # Number of data used for train, validation, and test
    rows = task_rows(TASK, split)
    print(f"{split}'s total rows: {len(rows)}")
    for source in sources:
        count = (rows["source"] == source).sum()
        print(f"{source:<22} {count:>6} rows")
    print()
```

## Check information on single sample

Load one PlayPanT val window and print its shapes.

```python
val_rows = task_rows(TASK, "val", sources=["playpant"])
print(f"Play pant val rows: {len(val_rows)}\n")

sample = load_sample(TASK, val_rows.iloc[0])
print(f"Clip id: {sample.clip_id}")
print(f"Audio shape: {sample.audio.shape}")  # 16kHz mono * 10 secs
print(f"Targets shape: {sample.targets.shape}")  # 250 segments of 40ms data, # of SED model classes
print(f"Has a dog: {sample.is_dog}")
```
