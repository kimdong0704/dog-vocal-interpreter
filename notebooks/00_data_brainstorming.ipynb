# 00 Data Brainstorming

**Purpose:** Record which datasets were considered for the project, which were kept, which were dropped, and which are left for later.

**Inputs:** None (planning notes and dataset links only).

**Outputs:** The dataset list behind `configs/datasets.yaml`; no files are written.

**Splits used:** None (no data is loaded; test: not used).

## 1. Kept (11 in `datasets.yaml`)

The datasets in `configs/datasets.yaml`, with size and why each is kept.

### Dog play pant (PlayPanT) | **0.38 GB**

* https://zenodo.org/records/18972388
* Only open set with dog IDs, expert onset/offset labels and no overlap with model pretraining
* Bark-detector validation; play slice for the audio branch

### AudioSet-Strong-Animals (eval) | **3.0 GB**

* https://huggingface.co/datasets/enyoukai/AudioSet-Strong-Animals
* Same label space and frame-level labels as the bark detector
* Secondary validation (optimistic)

### ESC-50 | **0.6 GB**

* https://github.com/karolpiczak/ESC-50
* Household non-dog sounds (doors, vacuum, cat, footsteps) as false-alarm negatives
* also noise for augmentation

### Barkopedia Activity & Environment | **0.61 GB**

* https://huggingface.co/datasets/ArlingtonCL2/Barkopedia_Dog_Act_Env
* Only large set with situation labels, covering 3 of our 4 situation categories, plus other
* Training only

### EmotionalCanines | **0.07 GB**

* https://huggingface.co/datasets/ArlingtonCL2/BarkopediaDogEmotionClassification_Data
* Only open arousal/valence dog audio, with dog-disjoint splits
* Emotion head

### Stanford Dogs (Kaggle mirror) | **0.76 GB**

* https://www.kaggle.com/datasets/jessicali9530/stanford-dogs-dataset
* The images that StanfordExtra, Dog-Pose and DogPoseCV label

### StanfordExtra (Kaggle mirror) | **0.02 GB**

* https://www.kaggle.com/datasets/sachadee/stanford-dogs-dataset-for-animal-pose-estimation
* Only labels with tail tip and ear tips across many breeds, MIT, with official splits

### Ultralytics Dog-Pose | **0.35 GB**

* https://docs.ultralytics.com/datasets/pose/dog-pose/
* Fastest pose baseline (images + keypoints in one zip)

### DogPoseCV | **0.82 GB**

* https://huggingface.co/datasets/stockeh/dog-pose-cv
* Clean human-labelled posture (standing/sitting/lying) for the body-language branch

### Roboflow ai-stray-dog-project v1 | **<1 GB**

* https://universe.roboflow.com/rusydan/ai-stray-dog-project
* Only sizeable source of leash boxes

### Roboflow cat-dog-bowl v20 | **<1 GB**

* https://universe.roboflow.com/eyedetection-s8fd9/cat-dog-bowl-ncx2x
* Bowl boxes in home-like phone footage

### **Total:** about 8 GB

## 2. Planned for download, then dropped

Datasets that were planned but dropped, with the reason.

### rmarcosg bark-detection | **1.93 GB**

* https://huggingface.co/datasets/rmarcosg/bark-detection
* Extra non-dog negatives (music, voice, claps, vacuum)
* Dropped: positives are one dog with human-imitated barks; negatives overlap ESC-50/FSD50K

### AP-10K | **~1–2 GB**

* https://github.com/AlexTheBad/AP-10K
* Only source of dog eye keypoints
* Dropped: no tail tip or ears, already in ViTPose++ pretraining, Google Drive download

### BADJA | **~2 GB**

* https://github.com/benjiebob/BADJA
* Tail-tip tracks over time, to sanity-check wag frequency
* Dropped: only 218 frames, needs a separate DAVIS download

### FSD50K (CC0/CC-BY subset) | **~2.5 GB**

* https://zenodo.org/records/4060432
* Largest permissive pool of hard negatives
* Cut: weak labels, per-clip licenses, many small files

### DEMAND | **1.77 GB**

* https://zenodo.org/records/1227121
* Real home-room noise for augmentation
* Cut: ESC-50 now covers noise

### COCO val2017 | **~1 GB**

* https://cocodataset.org
* Smoke test for the dog/person detector
* Cut: the detector's own benchmark, few home frames

### LVIS v1 val | **~0.1 GB**

* https://www.lvisdataset.org
* Collar and doorknob boxes for scene checks
* Cut: zip URL unverified, needs COCO images

### Open Images V7 (dog subset) | **~1–2 GB**

* https://storage.googleapis.com/openimages/web/factsfigures_v7.html
* Door, ball and bowl boxes outside detector pretraining
* Cut: ~3,000 separate files, no leash boxes

### Kinetics-700 dog classes | **~0.35 GB**

* https://github.com/cvdfoundation/kinetics-dataset
* Dog-with-person video for hard negatives and scene frames
* Cut: barking not guaranteed, auxiliary only

### Kaggle Dog Emotion | **0.16 GB**

* https://www.kaggle.com/datasets/danielshanbalico/dog-emotion
* Weak pretraining for body language
* Cut: one non-expert annotator, background shortcuts, never usable for evaluation*

### Kaggle Dog Emotions-5 | **0.23 GB**

* https://www.kaggle.com/datasets/dougandrade/dog-emotions-5-classes
* Same as above
* Same as above

## 3. To Be Implemented In The Future

Larger or later datasets that may be added in the future.

### Dog play pant raw video | **43.4 GB**

* https://zenodo.org/records/18972388
* Only synchronized audio + video with dog IDs (fusion check, play only)
* Very large; play only

### DogSpeak | **3.83 GB**

* https://huggingface.co/datasets/ArlingtonCL2/Dog-Vocal-Separation
* Noisy/clean bark pairs for denoising
* Stretch goal

### MUSAN | **11 GB**

* https://www.openslr.org/17/
* More noise for augmentation
* Large; only if budget remained

### VGGSound (mteb mirror) | **~0.4 GB kept**

* https://huggingface.co/datasets/mteb/VGGSound
* ~190 audio-visual dog clips
* Streams 19.7 GB for 0.4 GB kept; in audio-encoder pretraining

### DogFLW | **1.47 GB**

* https://github.com/martvelge/DogFLW
* Dog face and ear landmarks
* Optional; same images as StanfordExtra

### Yin & McCowan dog barks | **~1.56 GB**

* https://archive.org/details/dog_barks
* Best fit to stranger alert / separation / play, with dog IDs
* No author license; usable only with written permission from Brenda McCowan
