# 50 Fusion Dataset Cache


Fusion of audio, body and scene (CLIP) per bark segment, trained on the fusion collection

(60 train / 20 val / 20 test videos, split by dog). Test is scored once, at the end.

```python
import time
import pandas as pd

from wooftrainer.fusion_trainer.branch_outputs import cache_branch_outputs, load_branch_outputs
```

## Load Fusion Data and Cache It

```python
start = time.perf_counter()
cache_branch_outputs()
print(f"\ntook {(time.perf_counter() - start) / 60:.1f} min")
```

```python
segments, videos = load_branch_outputs()
print(len(segments), "segments from", len(videos), "videos\n")

print("videos with no segment in their span, per split and category:")
print(videos[videos["segments_kept"] == 0].groupby(["split", "category"]).size().to_string(), "\n")

print("segments per video:", videos["segments_kept"].describe().round(1).to_dict(), "\n")

presence = segments.groupby("category")[["audio_present", "body_present", "scene_present"]].mean()
print("share of segments where each branch is present:")
print(presence.round(2).to_string(), "\n")

print("segments per split and category:")
print(segments.groupby(["split", "category"]).size().unstack().to_string())
```
